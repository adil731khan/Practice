```python
sales=pd.DataFrame({
    "Salesperson":["A","A","B","B","C","C"],
    "City":["Delhi","Lucknow","Delhi","Kanpur","Delhi","Lucknow"],
    "Product":["Pen","Book","Pen","Book","Book","Pen"],
    "Sales":[120,250,150,300,400,180]}
)
sales
```

```python
sales["Sales"].sum()
```

```python
sales.groupby("Salesperson")["Sales"].sum()
```

```python
sales.groupby("City")["Sales"].sum()
```

```python
sales.pivot_table(values="Sales",index="City",columns="Product",aggfunc="sum")
```
